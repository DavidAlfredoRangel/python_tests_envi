```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Datos históricos (2015 - 2025) por cada 100,000 habitantes
datos = {
    "Año": [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024, 2025],
    "Tasa_Muerte": [545.0, 555.0, 563.0, 574.0, 588.0, 860.0, 879.0, 659.0, 619.0, 630.0, 606.0],
    "Tasa_Causas_Externas": [53.20, 55.80, 60.10, 63.40, 62.90, 61.50, 63.80, 61.20, 60.50, 61.80, 57.40],
    "Tasa_Asesinato": [17.00, 20.00, 25.00, 29.00, 29.00, 29.00, 28.00, 25.00, 24.80, 25.80, 21.40],
    "Tasa_Homicidio_Doloso": [13.17, 16.30, 20.06, 23.09, 23.19, 22.49, 21.92, 20.23, 19.30, 19.45, 16.80],
    "Tasa_Homicidio_Culposo": [13.85, 13.40, 13.10, 12.60, 12.15, 10.90, 11.45, 12.30, 12.80, 12.65, 12.10],
    "Tasa_Suicidio": [5.10, 5.10, 5.20, 5.30, 5.60, 6.20, 6.50, 6.30, 6.90, 6.80, 6.70],
    "Tasa_Feminicidio": [0.66, 0.96, 1.17, 1.41, 1.46, 1.45, 1.49, 1.45, 1.28, 1.21, 1.05]
}

# 2. Crear la tabla y mostrarla
df = pd.DataFrame(datos)
print("=== TABLA DE TASAS (2015 - 2025) ===")
print(df.to_string(index=False))

# 3. Resumen estadístico rápido (promedio, mínimo y máximo)
print("\n=== RESUMEN ESTADÍSTICO ===")
print(df.drop(columns="Año").describe().loc[["mean", "min", "max"]].round(2))

# 4. Graficar en 3 paneles simples según su escala
fig, ax = plt.subplots(3, 1, figsize=(10, 11), sharex=True)

# Gráfica 1: Mortalidad general
ax[0].plot(df["Año"], df["Tasa_Muerte"], marker="o", color="navy", label="Tasa de muerte")
ax[0].set_title("1. Tasa de muerte general")
ax[0].set_ylabel("Por 100 mil hab.")
ax[0].grid(True, alpha=0.3)
ax[0].legend()

# Gráfica 2: Causas externas y homicidios
ax[1].plot(df["Año"], df["Tasa_Causas_Externas"], marker="o", label="Causas externas")
ax[1].plot(df["Año"], df["Tasa_Asesinato"], marker="s", label="Asesinato")
ax[1].plot(df["Año"], df["Tasa_Homicidio_Doloso"], marker="^", label="Homicidio doloso")
ax[1].plot(df["Año"], df["Tasa_Homicidio_Culposo"], marker="d", label="Homicidio culposo")
ax[1].set_title("2. Causas externas y homicidios")
ax[1].set_ylabel("Por 100 mil hab.")
ax[1].grid(True, alpha=0.3)
ax[1].legend()

# Gráfica 3: Suicidio y feminicidio
ax[2].plot(df["Año"], df["Tasa_Suicidio"], marker="o", color="purple", label="Suicidio")
ax[2].plot(df["Año"], df["Tasa_Feminicidio"], marker="s", color="crimson", label="Feminicidio")
ax[2].set_title("3. Suicidio y feminicidio")
ax[2].set_xlabel("Año")
ax[2].set_ylabel("Por 100 mil hab.")
ax[2].set_xticks(df["Año"])
ax[2].grid(True, alpha=0.3)
ax[2].legend()

plt.tight_layout()
plt.show()
```
